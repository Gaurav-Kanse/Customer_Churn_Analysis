```python
from scipy.stats import chi2_contingency

contract_table = pd.crosstab(
    df_clean["Contract"],
    df_clean["Churn"]
)

chi2, p_value, dof, expected = chi2_contingency(
    contract_table
)

print("Contingency Table:")
display(contract_table)

print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.10f}")

if p_value < 0.05:
    print("Result: Statistically significant association detected.")
else:
    print("Result: Insufficient evidence of an association.")
```

```python
internet_table = pd.crosstab(
    df_clean["InternetService"],
    df_clean["Churn"]
)

chi2, p_value, dof, expected = chi2_contingency(
    internet_table
)

print("Contingency Table:")
display(internet_table)

print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.10f}")

if p_value < 0.05:
    print("Result: Statistically significant association detected.")
else:
    print("Result: Insufficient evidence of an association.")
```

```python
from scipy.stats import chi2_contingency

def cramers_v(table):
    chi2, p_value, dof, expected = chi2_contingency(table)

    n = table.to_numpy().sum()
    rows, cols = table.shape

    denominator = min(rows - 1, cols - 1)

    if denominator == 0:
        return np.nan

    v = np.sqrt(chi2 / (n * denominator))

    return v

contract_table = pd.crosstab(
    df_clean["Contract"],
    df_clean["Churn"]
)

contract_v = cramers_v(contract_table)

internet_table = pd.crosstab(
    df_clean["InternetService"],
    df_clean["Churn"]
)

internet_v = cramers_v(internet_table)


print(f"Cramér's V - Contract Type vs Churn: {contract_v:.4f}")
print(f"Cramér's V - Internet Service vs Churn: {internet_v:.4f}")
```

```python
numerical_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

outlier_results = []

for column in numerical_columns:

    values = df_clean[column].dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (values < lower_bound) |
        (values > upper_bound)
    )

    outlier_results.append({
        "Variable": column,
        "Q1": round(q1, 2),
        "Q3": round(q3, 2),
        "IQR": round(iqr, 2),
        "Lower_Bound": round(lower_bound, 2),
        "Upper_Bound": round(upper_bound, 2),
        "Potential_Outliers": int(outlier_mask.sum()),
        "Outlier_Percentage": round(
            outlier_mask.mean() * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

```python
demographic_columns = [
    "SeniorCitizen",
    "Dependents",
    "Partner"
]

for column in demographic_columns:

    summary = (
        df_clean.groupby(column, observed=True)
        .agg(
            Total_Customers=("Churn", "size"),
            Churned_Customers=(
                "Churn",
                lambda x: (x == "Yes").sum()
            )
        )
    )

    summary["Churn_Rate_%"] = (
        summary["Churned_Customers"]
        / summary["Total_Customers"]
        * 100
    ).round(2)

    print(f"\nChurn by {column}:")
    display(summary)
```

```python
summary_segments = {
    "Contract": "Month-to-month",
    "InternetService": "Fiber optic",
    "PaymentMethod": "Electronic check",
    "Dependents": "No",
    "Partner": "No",
    "SeniorCitizen": 1
}

for column, category in summary_segments.items():

    segment = df_clean[df_clean[column] == category]

    total = len(segment)

    churned = (segment["Churn"] == "Yes").sum()

    churn_rate = (
        churned / total * 100
        if total > 0 else 0
    )

    print(f"\n{column}: {category}")
    print(f"Total customers: {total}")
    print(f"Churned customers: {churned}")
    print(f"Churn rate: {churn_rate:.2f}%")
```
